```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
```

```python
df = pd.read_csv('jfk_dummy_jan25.csv')
df.head(20)
```

```python
df.info()
```

**Set the dataset**

```python
x = df[['temperature', 'wind_speed', 'rain']]
y = df['delayed']
```

**Split the dataset**

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state=3)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

**Logistic Regression Model**

```python
lr_model = LogisticRegression()
lr_model.fit(x_train, y_train)
```

**Make predictions** (*the data is very basic, so not the best example!* )

```python
y_pred = lr_model.predict(x_test)
```

```python
print(y_pred)
```

**Decision Tree Classifier**

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt_model = DecisionTreeClassifier()
dt_model.fit(x_train, y_train)
```

```python
y_pred_dt = dt_model.predict(x_test)
```

**Find accuracy score**

```python
print('Accuracy Score:', metrics.accuracy_score(y_test, y_pred_dt)) #test dataset is very basic, so accuracy score will be low
```

**Confusion Matrix**

```python
print(metrics.confusion_matrix(y_test, y_pred_dt)) #unsure if this is correct?!
```
