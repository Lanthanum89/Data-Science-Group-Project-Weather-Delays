Florida January 2025

```python
import pandas as pd
import matplotlib.pyplot as plt
florida = pd.read_csv('FloridaJanData.csv')
```

```python
florida.head()  
```

```python
florida.info
```

```python
florida.describe()
```

```python
plt.figure(figsize=(10,6))
florida['EVENT_TYPE'].value_counts().plot(kind='bar')
plt.title('Frequency of Event Types in Florida (Jan 2025)')
plt.xlabel('Event Type')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
event_counts = florida['EVENT_TYPE'].value_counts()
plt.figure(figsize=(8,8))
plt.pie(event_counts, labels=event_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Proportion of Event Types in Florida (Jan 2025)')
plt.tight_layout()
plt.show()
```
