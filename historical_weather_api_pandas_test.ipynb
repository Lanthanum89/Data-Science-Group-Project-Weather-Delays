## Function for calling the Historical Weather API

```python
def get_weather_info(row):

  latitude = row['LATITUDE']
  longitude = row['LONGITUDE']
  start_date = '2015-01-01'
  end_date = '2015-01-01'

  endpoint = 'https://archive-api.open-meteo.com/v1/archive?latitude={}&longitude={}&start_date={}&end_date={}&daily=weather_code&timezone=auto'.format(latitude, longitude, start_date, end_date)

  response = requests.get(url=endpoint)
  data = response.json()
  weather_codes = data['daily']['weather_code']

  row['API_WEATHER_CODES'] = weather_codes
  return row
```

## Apply the API function row wise

```python
flights_df_merged = flights_df_merged.apply(get_weather_info, axis=1)
```
