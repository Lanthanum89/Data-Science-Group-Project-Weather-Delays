```python
df['weather_pct'] = ((df['weather_ct'] / df['arr_del15']) * 100).round(2)
df['weather_pct']
```

```python
import seaborn as sns

sns.histplot(df['weather_pct'], bins=20, kde=True)
plt.xlabel('Weather Delay Percentage')
plt.title('Distribution of Weather Delay Percentages')
plt.show()
```

```python
import seaborn as sns

sns.histplot(df['arr_del15'], bins=50, kde=True)
plt.xlabel('Flights delayed more than 15 minutes by airport')
plt.title('Flights delayed more than 15 minutes by airport')
plt.show()
```

```python
df.isna().sum()

There is minimal missing data in this data set
```

```python
df['delayed'] = (df['arr_del15'] > 0).astype(int)
df['weather_pct'] = df['weather_ct'] / df['arr_del15']
df['carrier_pct'] = df['carrier_ct'] / df['arr_del15']
```

```python
import seaborn as sns
sns.histplot(df['weather_delay'], bins=80)


```

```python
sns.boxplot(x='delayed', y='weather_delay', data=df)
```

```python
df.groupby('month')['weather_delay'].mean().plot(kind='bar')
```

```python
sns.boxplot(x='month', y='weather_delay', data=df)
plt.title('Weather Delays by Month')
```

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=False, cmap='coolwarm')

```

```python
top_airports = df.groupby('airport')['weather_delay'].mean().sort_values(ascending=False).head(10)
top_airports.plot(kind='bar')
plt.title('Top Airports by Average Count Weather Delay')
```

```python
top_airports = df.groupby('airport')['weather_pct'].mean().sort_values(ascending=False).head(10)
top_airports.plot(kind='bar')
plt.title('Top Airports by Average Percentage Weather Delay')
```
